```python
# HERE WE WILL SORT DATA AND USE THE PERCENTILE FOR FIND Q1 & Q3.
sorting=df["market_price"].sort_values(ascending=True)
print(sorting)
# WE FOUND Q1 AND Q3.
a=np.percentile(sorting,[25,50,75,100])
(f"this is data in percentile:{a}")
```

```python
# HERE WE WILL FIND {IQR}.
Q3=425
Q1=100
IQR=Q3-Q1
print(IQR)
```

```python
# HERE WE WILL DO SOME OPERATIONS FOR FINDING A RANGE BY WHICH WE CAN FIND OUTLIERS.
Min_range=(1.5*IQR)-Q1
Max_range=(1.5*IQR)+Q3
print(Min_range)
print(Max_range)
```

```python
# HERE WE USE GROUPBY FOR FIND COUNT OF MARKET_PRICE.

d=df.groupby("market_price").size().reset_index(name="amount_count")
d
```

```python
# HERE BY THE RANGE WE WILL FIND OUT OUTLIERS.
h=d[d["market_price"]>Max_range]
print(h)
```

```python
# HERE BY THE RANGE WE WILL FIND OUT OUTLIERS.
f=d[d["market_price"]<Min_range]
print(f)
```

```python
market_outliers=(h,f)
market_outliers
```

```python
# NOW HERE WE EXTRACT MEAN & CHANGE DATATYPE.
mean_1=df["market_price"].sort_values().mean().astype(int)
print(mean_1)
```

```python
# NOW HERE WE CAN LOOK PRODUCTS WITH PRICE WHICH IS MORE THEN A RANGE OR LESS THEN RANGE.
outliers = df[(df["market_price"] > Max_range) | (df["market_price"] < Min_range)]
print(outliers[["product","market_price"]])
```

```python
# NOW HERE WE WILL REPLACE THE OUTLIERS INTO MEAN.

df.loc[(df["market_price"] > Max_range) | (df["market_price"] < Min_range),"market_price"]=mean_1
df
```

```python
# NOW HERE WE WILL USE QUANTILE FOR FIND OUT QTR1 & QTR3 AND EXTRACT IQR.
q1=df["sale_price"].quantile(0.25)
q3=df["sale_price"].quantile(0.75)
IQR=q3-q1
print(q1)
print(q3)
print(IQR)
```

```python
# HERE WE WILL USE SOMEMETHOD FOR FETCHING RANGE BY WHICH WE WILL FIND OUTLIERS.
range1=(IQR*1.25)-q1
range2=(IQR*1.25)+q3
print(range1)
print(range2)
```

```python
# HERE WE USE GROUPBY ON SALE_PRICE AND USE SIZE DOR FIND COUNT OF SALE_PRICE.
d=df.groupby("sale_price").size().reset_index(name="prize_count")
print(d)
```

```python
# HERE BY THE RANGE WE WILL FIND OUTLIERS.
h=d[d["sale_price"]<range1]
print(h)
```

```python
# HERE BY THE RANGE WE WILL FIND OUTLIERS.
h2=d[d["sale_price"]>range2]
print(h2)
```

```python
# HERE WE WILL EXTRACT MEAN FROM DATASET AND CHANGE DATATYPE.
mean2=df["sale_price"].sort_values().mean().astype(int)
print(mean2)
```

```python
# HERE WE WILL TAKE OVER LOOK ON PRODUCT AND SALE_PRICE ON BEHALF OF RANGE.
outlier=df[(df["sale_price"] < range1) | (df["sale_price"] > range2)]
print(outlier[["product","sale_price"]])
```

```python
#HERE WE WILL REPLACE OUTLIERS INTO MEAN.
df.loc[(df["sale_price"] < range1) | (df["sale_price"] > range2),"sale_price"]=mean2
df
```
