```python
plt.figure(figsize=(19,9))
sns.histplot(df["Age"],kde=True)
plt.title("DISTRIBUTION OF AGE")
plt.show()
```

```python
plt.figure(figsize=(8,7))
sns.histplot(df["Annual_Premium"],kde=True,bins=30)
plt.title("Distribution of Annual_Premium")
plt.show()
```

```python
plt.figure(figsize=(5,5))
a=df["Response"].value_counts()
plt.pie(a,labels=a.index,autopct="%1.1f%%",startangle=90,wedgeprops=dict(width=0.4))
plt.title("Claim Frequency(0=NO , 1= YES)")
plt.legend(a.index)
plt.plot()
```

```python
plt.figure(figsize=(8,8))
sns.countplot(data=df,x="Gender",hue="Response")
plt.title("Claims by Gender")
plt.show()
```

```python
plt.figure(figsize=(9,9))
sns.countplot(data=df,x="Vehicle_Age",hue="Response")
plt.title("Claims by Vehicle Age")
plt.show()
```

```python
plt.figure(figsize=(12,12))
filtered = df[df["Response"] == 1]

sns.countplot(data=filtered, y="Region_Code")
plt.title("Customer regions")
plt.show()
```

```python
plt.figure(figsize=(12,8))
a = df["Policy_Sales_Channel"].value_counts()

top_channels = a.head(20)
plt.barh(top_channels.index.astype(str), top_channels.values)
plt.xlabel("Number of Policies")
plt.ylabel("Policy Sales Channel")
plt.title("Top 20 Distribution of Policy Sales Channels")
plt.show()
```

```python
plt.figure(figsize=(6,6))
sns.countplot(data=df, x="Vehicle_Damage", hue="Response")
plt.title("Claims by Vehicle Damage")
plt.show()
```

```python
plt.figure(figsize=(10,8))
top_20_channels = df["Policy_Sales_Channel"].value_counts().head(20).index
filtered_df_channels = df[df["Policy_Sales_Channel"].isin(top_20_channels)]
sns.countplot(data=filtered_df_channels, y="Policy_Sales_Channel", hue="Response", order=top_20_channels)
plt.title("Claim by Top 20 Policy Sales Channels")
plt.xlabel("Count")
plt.ylabel("Policy Sales Channel")
plt.show()
```
